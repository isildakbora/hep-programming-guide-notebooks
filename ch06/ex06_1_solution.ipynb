Exercise 6.1 MC integration of σ(e⁺e⁻ → μ⁺μ⁻) revised Implement a simple Monte Carlo integration to compute the total cross section for \(e^+e^- \to \mu^+\mu^-\) at \(\sqrt{s} = 10\) GeV, well below the Z pole, by integrating the QED differential cross section \(d\sigma/d\cos\theta = \frac{\pi\alpha^2}{2s}(1+\cos^2\theta)\). Compare with the analytical result \(\sigma = \frac{4\pi\alpha^2}{3s}\). Optionally, compare the QED formula evaluated at \(\sqrt{s} = 91.2\) GeV with the measured peak cross section at the Z pole (about 2 nb for \(\mu^+\mu^-\)) and explain the difference.

<details><summary>Hint</summary>

Sample \(\cos\theta\) uniformly in \([-1, 1]\) (volume 2; the \(\phi\) integral gives \(2\pi\) if you start from \(d\sigma/d\Omega\)). Convert from natural units with \((\hbar c)^2 = 0.3894\) mb GeV\(^2\) \(= 3.894\times10^{5}\) nb GeV\(^2\). At the Z pole the process goes through the Z resonance, which the pure photon-exchange formula ignores.

</details>

```python
import sys
IN_COLAB = "google.colab" in sys.modules
if IN_COLAB:
    %pip install -q numpy matplotlib
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(42)
```

### 1. Sample the scattering angle and integrate

Use the volume factor 2 for uniform cos θ on [−1,1]. Convert GeV⁻² to nb only after integration.

```python
alpha = 1/137
s = 10.0**2  # GeV^2
conv = 3.894e5  # nb GeV^2
n_samples = 100_000  # increase for more precision
cos_theta = rng.uniform(-1, 1, n_samples)
dsigma_dc = np.pi * alpha**2 / (2*s) * (1+cos_theta**2)
sigma_mc = 2 * dsigma_dc.mean() * conv
mc_se = 2 * dsigma_dc.std(ddof=1) / np.sqrt(n_samples) * conv
sigma_exact = 4*np.pi*alpha**2/(3*s)*conv
print(f"MC: {sigma_mc:.5f} ± {mc_se:.5f} nb; analytic: {sigma_exact:.5f} nb")
assert abs(sigma_mc-sigma_exact)<5*mc_se
```

### 2. Inspect the angular dependence and the Z-pole comparison

```python
c_grid = np.linspace(-1,1,300)
fig,ax=plt.subplots()
ax.plot(c_grid, np.pi*alpha**2/(2*s)*(1+c_grid**2)*conv)
ax.set(xlabel=r"$\cos\theta$",ylabel=r"$d\sigma/d\cos\theta$ [nb]",title=r"$e^+e^-\to\mu^+\mu^-$ at 10 GeV")
plt.show()
sigma_z_photon = 4*np.pi*alpha**2/(3*91.2**2)*conv
print(f"Pure photon at 91.2 GeV: {sigma_z_photon:.4f} nb")
```

The result is about **0.87 nb** at 10 GeV, in agreement with the book. The photon-only estimate near 91.2 GeV is about 0.010 nb, roughly 200 times below the ≈2 nb Z-pole cross section because resonant Z exchange is absent from this QED calculation.

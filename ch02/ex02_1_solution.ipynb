Exercise 2.1 Rapidity Extend the Particle class with a method rapidity() that computes the rapidity \(y = \frac{1}{2} \ln \frac{E + p_z}{E - p_z}\). Add appropriate error handling for edge cases.

<details><summary>Hint</summary>

The formula breaks down when \(E \le |p_z|\), which happens for a massless particle along the beam or through rounding. Decide what to return in that case (a large signed value, as eta() does, or an exception), and check that rapidity() agrees with eta() for a massless particle.

</details>

```python
import subprocess, shutil, sys
IN_COLAB = "google.colab" in sys.modules
assert shutil.which("g++"), "A C++ compiler (g++) is required"
```

### Implement and test boundary cases

```python
from pathlib import Path
source = r'''#include <cmath>
#include <stdexcept>
class Particle {
  double px_, py_, pz_, E_;
public:
  Particle(double px,double py,double pz,double E):px_(px),py_(py),pz_(pz),E_(E){}
  double px() const {return px_;} double py() const {return py_;}
  double pz() const {return pz_;} double energy() const {return E_;}
  double pt() const {return std::hypot(px_,py_);}
  double eta() const {return std::asinh(pz_/pt());}
  double rapidity() const {
    if (!std::isfinite(E_) || !std::isfinite(pz_) || E_ < std::abs(pz_))
      throw std::domain_error("Unphysical four-vector");
    if (E_ == std::abs(pz_)) return std::copysign(INFINITY,pz_);
    return 0.5*std::log((E_+pz_)/(E_-pz_));
  }
};
#include <iostream>
int main(){ Particle p(30,40,20,std::sqrt(2900.0));
  std::cout << "eta=" << p.eta() << " y=" << p.rapidity() << "\n";
  if (std::abs(p.eta()-p.rapidity()) > 1e-12) return 1;
  Particle beam(0,0,10,10); if (!std::isinf(beam.rapidity())) return 2;
  try { Particle bad(0,0,10,9); bad.rapidity(); return 3; }
  catch(const std::domain_error&) { std::cout << "invalid input caught\n"; }
}'''
Path("exercise.cpp").write_text(source)
build = subprocess.run(["g++", "-O2", "-std=c++17", "exercise.cpp", "-o", "exercise"], capture_output=True, text=True)
print(build.stderr)
assert build.returncode == 0
run = subprocess.run(["./exercise"], capture_output=True, text=True, check=True)
print(run.stdout)
```

### Interpretation

For a massless four-vector with nonzero transverse momentum, rapidity and pseudorapidity agree. A lightlike beam particle has infinite signed rapidity; unphysical energy throws.

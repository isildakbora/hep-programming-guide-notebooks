Exercise 2.2 A generic filter Write a template function filter that takes a container and a predicate (a function returning bool ) and returns a new container containing only elements that satisfy the predicate. Test it by selecting jets with \(p_T &gt; 30\) GeV.

<details><summary>Hint</summary>

Use template &lt;typename Container, typename Pred&gt; , create an empty Container result; and fill it with std::copy_if(c.begin(), c.end(), std::back_inserter(result), pred) (header &lt;iterator&gt; ). Pass the predicate as a lambda.

</details>

```python
import subprocess, shutil, sys
IN_COLAB = "google.colab" in sys.modules
assert shutil.which("g++"), "A C++ compiler (g++) is required"
```

### Generic selection

```python
from pathlib import Path
source = r'''#include <algorithm>
#include <iterator>
#include <iostream>
#include <vector>
struct Jet {double pt;};
template<typename Container,typename Pred>
Container filter(const Container& c, Pred pred) {
  Container result;
  std::copy_if(c.begin(), c.end(), std::back_inserter(result), pred);
  return result;
}
int main(){std::vector<Jet> jets{{20},{31},{45},{30}};
  auto selected=filter(jets,[](const Jet& j){return j.pt>30;});
  std::cout << "selected=" << selected.size() << "\n";
  return selected.size()==2 ? 0:1;
}'''
Path("exercise.cpp").write_text(source)
build = subprocess.run(["g++", "-O2", "-std=c++17", "exercise.cpp", "-o", "exercise"], capture_output=True, text=True)
print(build.stderr)
assert build.returncode == 0
run = subprocess.run(["./exercise"], capture_output=True, text=True, check=True)
print(run.stdout)
```

```python
assert "selected=2" in run.stdout
```

### Interpretation

The strict threshold selects the 31 and 45 GeV jets; 30 GeV is excluded.

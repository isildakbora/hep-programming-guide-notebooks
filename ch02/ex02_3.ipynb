Exercise 2.3 An Event class Implement an Event class that contains vectors of Particle objects (electrons, muons, jets) and provides methods to compute \(H_T\) (scalar sum of jet \(p_T\)), missing transverse energy, and the number of b-tagged jets.

<details><summary>Hint</summary>

\(H_T\) is a std::accumulate over jet \(p_T\) ( Listing 2.11 ). For \(E_T^{\mathrm{miss}}\), sum \(p_x\) and \(p_y\) over all visible objects and take the magnitude of the negative sum ( Eq. 1.7 ). Particle has no b-tag score, so store one per jet in a parallel vector or a small Jet struct.

</details>

### Build an event with three derived quantities

```python
from pathlib import Path
source = r'''#include <cmath>
#include <iostream>
#include <numeric>
#include <vector>
struct Particle {double px,py; double pt() const {return std::hypot(px,py);}};
struct Jet {Particle p; double btag_score;};
class Event {
public:
 std::vector<Particle> electrons,muons;
 std::vector<Jet> jets;
 double ht() const {// TODO: scalar sum of jet pt
    return 0;}
 double met() const {double px=0,py=0;
    for(const auto& p:electrons){px+=p.px;py+=p.py;}
    for(const auto& p:muons){px+=p.px;py+=p.py;}
    for(const auto& j:jets){px+=j.p.px;py+=j.p.py;}
    // TODO: return magnitude of negative visible pT sum
    return 0;}
 int n_bjets(double threshold=0.5) const {int n=0;
    for(const auto& j:jets) n+=j.btag_score>threshold;
    // TODO: count jets above threshold
    return n;}
};
int main(){Event e; e.electrons={{10,0}};e.muons={{0,20}};
 e.jets={{{30,40},0.7},{{-10,0},0.2}};
 std::cout << "HT=" << e.ht() << " MET=" << e.met()
           << " bjets=" << e.n_bjets() << "\n";
 return 0;
}'''
Path("exercise.cpp").write_text(source)
build = subprocess.run(["g++", "-O2", "-std=c++17", "exercise.cpp", "-o", "exercise"], capture_output=True, text=True)
print(build.stderr)
assert build.returncode == 0
run = subprocess.run(["./exercise"], capture_output=True, text=True, check=True)
print(run.stdout)
```

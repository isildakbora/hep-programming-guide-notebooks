Exercise 1.3 Data rates and trigger reduction Research the data rates at the LHC. If the LHC produces 40 million collisions per second and each collision produces approximately 1 MB of raw detector data, calculate the total raw data rate in petabytes per second. Compare this with the rate actually stored to permanent storage (\(\sim 1\) kHz after triggers). What is the overall data reduction factor?

<details><summary>Hint</summary>

Rate × event size gives bytes per second. Use \(1\ \mathrm{PB} = 10^{15}\) bytes and \(1\ \mathrm{MB} = 10^{6}\) bytes.

</details>

```python
import sys
IN_COLAB = "google.colab" in sys.modules
```

### Calculate the input and stored rates

```python
collision_rate_hz = 40e6
stored_rate_hz = 1e3
event_bytes = 1e6
raw_pb_s = collision_rate_hz * event_bytes / 1e15
stored_gb_s = stored_rate_hz * event_bytes / 1e9
reduction = collision_rate_hz / stored_rate_hz
print(f"Raw: {raw_pb_s:.2f} PB/s = {raw_pb_s*1000:.0f} TB/s")
print(f"Stored: {stored_gb_s:.1f} GB/s; reduction: {reduction:.0f}×")
assert (raw_pb_s,stored_gb_s,reduction)==(0.04,1.0,40000.0)
```

### Interpretation

At fixed event size, the event-rate and byte-rate reduction factors are the same. These are the exercise’s rounded assumptions, not a current machine-operations measurement.

**Book answer:** Raw: \(4 \times 10^{7}\ \mathrm{s^{-1}} \times 1\ \mathrm{MB} = 40\ \mathrm{TB/s} = 0.04\ \mathrm{PB/s}\). Stored: \(10^{3}\ \mathrm{s^{-1}} \times 1\ \mathrm{MB} = 1\ \mathrm{GB/s}\). The reduction factor in rate (and hence in data volume at fixed event size) is \(4 \times 10^{7} / 10^{3} = 4 \times 10^{4}\).

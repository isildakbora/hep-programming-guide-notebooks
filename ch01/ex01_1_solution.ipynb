### Generate and draw

```python
from pathlib import Path
Path("exp_macro.C").write_text(r'''#include <TCanvas.h>
#include <TH1F.h>
#include <TRandom3.h>
void exp_macro() {
  TRandom3 rng(42);
  TH1F h("h", "Exponential sample;x;Events / 0.1", 100, 0, 10);
  for (int i=0; i<10000; ++i) h.Fill(rng.Exp(2.0));
  TCanvas c("c", "Exponential", 800, 600);
  h.Draw(); c.SaveAs("exp.pdf");
  printf("Sample mean: %.3f\n", h.GetMean());
}''')
ROOT.gROOT.ProcessLine(".x exp_macro.C")
```

```python
from pathlib import Path
assert Path("exp.pdf").is_file()
```

### Interpretation

The seeded sample mean should be close to 2. The histogram stops at 10, so its displayed mean is slightly below the full-sample mean.

Exercise 1.2 Counting Z bosons revised Using Equation (1.1) , calculate the expected number of \(Z \to e^+e^-\) and \(Z \to \mu^+\mu^-\) decays (combined) at the LHC with \(\sqrt{s} = 13.6\) TeV, given the inclusive production cross section \(\sigma(pp \to Z) \approx 60\) nb, the branching fraction \(\mathcal{B}(Z \to \ell^+\ell^-) \approx 3.37\%\) per lepton flavor, and \(\mathcal{L}_{\mathrm{int}} = 150\ \mathrm{fb^{-1}}\). Ignore detector acceptance. Express your answer in millions of events.

<details><summary>Hint</summary>

Convert everything to femtobarns first: \(1\ \mathrm{nb} = 10^{3}\ \mathrm{pb} = 10^{6}\ \mathrm{fb}\). Then multiply by the luminosity and by the branching fraction for two lepton flavors.

</details>

```python
import sys
IN_COLAB = "google.colab" in sys.modules
```

### Convert units and count decays

```python
sigma_nb = 60.0
lumi_fb_inv = 150.0
branching_per_flavor = 0.0337
sigma_fb = sigma_nb * 1e6
n_z = sigma_fb * lumi_fb_inv
n_millions = n_z * (2 * branching_per_flavor) / 1e6
print(f"{n_millions:.1f} million ee + μμ decays")
assert abs(n_millions - 606.6) < 0.1
```

### Interpretation

The expected yield is 606.6 million, or about 600 million, before acceptance and selection.

**Book answer:** \(60\ \mathrm{nb} = 6.0 \times 10^{4}\ \mathrm{pb} = 6.0 \times 10^{7}\ \mathrm{fb}\). Multiplying by \(150\ \mathrm{fb^{-1}}\) gives \(9.0 \times 10^{9}\) Z bosons. With \(\mathcal{B}(ee) + \mathcal{B}(\mu\mu) \approx 2 \times 0.0337 = 0.0674\), \(N \approx 9.0 \times 10^{9} \times 0.0674 \approx 6.1 \times 10^{8}\), i.e. about 600 million dilepton decays before acceptance and selection.

Exercise 4.3 Variable binning for a transverse mass Implement variable binning for a histogram of the transverse mass \(m_T\) in the \(H \to WW\) channel, using finer bins in the region below \(m_H\), where the distribution rises to its kinematic edge, and coarser bins in the tails.

<details><summary>Hint</summary>

For \(H \to WW \to \ell\nu\ell\nu\), the transverse mass cannot exceed \(m_H = 125.20\)&nbsp;GeV (up to resolution effects), so most signal lies between roughly 60 and 130&nbsp;GeV. Build the bin-edge array as in Listing 4.3 and remember to divide by the bin width before plotting.

### Generate toy events and normalize by bin width

```python
import numpy as np
import matplotlib.pyplot as plt
rng=np.random.default_rng(42)
m_H=125.20  # GeV, value used in the book
N=20000
truth=m_H*rng.beta(3.5,1.8,N)
m_T=truth+rng.normal(0,3,N)  # toy detector resolution
edges=np.r_[np.arange(0,60,10),np.arange(60,132,4),np.arange(140,201,20)]
edges=np.unique(edges)
counts,_=np.histogram(m_T,bins=edges)
density=counts/np.diff(edges)
fig,ax=plt.subplots(figsize=(8,5))
ax.stairs(density,edges,color='#0072B2',linewidth=2,label='Toy H → WW')
ax.axvline(m_H,color='#D55E00',ls='--',label=r'$m_H=125.20$ GeV')
ax.set_xlabel(r'$m_T$ [GeV]');ax.set_ylabel('Events / GeV')
ax.legend(frameon=False);fig.savefig('variable_mt.png',dpi=150,bbox_inches='tight')
plt.show()
print('Bin widths [GeV]:',np.unique(np.diff(edges)))
```

```python
assert len(counts)==len(edges)-1 and np.all(np.diff(edges)>0)
```

### Interpretation

Most toy signal appears below the 125.20 GeV edge; resolution can put reconstructed values above it. Width normalization makes heights comparable across unequal bins.
